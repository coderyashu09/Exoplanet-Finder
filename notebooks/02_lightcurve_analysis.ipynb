# 02 — Stellar Light Curve Processing & Phase Folding
Photometry retrieval, transit masking, Savitzky-Golay detrending, and dual-view generation.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import numpy as np
import matplotlib.pyplot as plt
from src.config import cfg
from src.lightcurve_downloader import LightCurveCache, download_stellar_lightcurve
from src.preprocessing import validate_and_align_time_system, flatten_lightcurve
from src.phase_folding import generate_two_views
```

```python
cache = LightCurveCache()
time_arr, flux_arr, flux_err, meta = download_stellar_lightcurve(11442793, cache=cache)
print(f'Loaded {len(time_arr)} cadences for Kepler-90 across {meta.get("quarters")} quarters.')
```

```python
t_clean = validate_and_align_time_system(time_arr)
flat_flux, trend, f_meta = flatten_lightcurve(t_clean, flux_arr, period=14.449, t0=133.0, duration_hours=3.5)
global_v, local_v, v_meta = generate_two_views(t_clean, flat_flux, period=14.449, t0=133.0, duration_hours=3.5)
print('Global View Shape:', global_v.shape)
print('Local View Shape:', local_v.shape)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
axes[0].plot(global_v, color='#63B3FF', lw=1)
axes[0].set_title('Global View (2001 bins)')
axes[1].plot(local_v, color='#35D49A', lw=1.5)
axes[1].set_title('Local View (201 bins)')
plt.tight_layout()
plt.show()
```
