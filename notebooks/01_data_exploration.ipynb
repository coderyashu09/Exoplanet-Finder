# 01 — NASA Kepler DR24 TCE Data Exploration
Initial exploratory data analysis on Kepler Q1-Q17 DR24 Threshold Crossing Events.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from src.config import cfg
from src.data_loader import get_or_build_tce_catalog, clean_and_filter_tce_data
```

```python
df_raw = get_or_build_tce_catalog()
print('Total raw records:', len(df_raw))
df_raw.head()
```

```python
df_clean, audit = clean_and_filter_tce_data(df_raw)
print('Cleaned records:', len(df_clean))
print('Unique Kepler stars:', df_clean['kepid'].nunique())
print(df_clean['av_training_set'].value_counts())
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
df_clean['tce_period'].hist(bins=50, ax=axes[0], log=True, color='#63B3FF')
axes[0].set_title('Orbital Period [days]')
df_clean['tce_duration'].hist(bins=50, ax=axes[1], color='#35D49A')
axes[1].set_title('Transit Duration [hrs]')
df_clean['tce_depth'].hist(bins=50, ax=axes[2], log=True, color='#FFBE55')
axes[2].set_title('Transit Depth [ppm]')
plt.tight_layout()
plt.show()
```
