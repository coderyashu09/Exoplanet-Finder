# 03 — Model Experiments: Baseline vs Two-View CNN
Compares tabular HistGradientBoosting baseline against Two-View 1D Convolutional Neural Network.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
import torch
from src.config import cfg
from src.baseline import BaselinePipeline
from src.model import DualViewExoplanetCNN
```

```python
df_train = pd.read_csv(cfg.processed_dir / 'tce_train.csv')
df_val = pd.read_csv(cfg.processed_dir / 'tce_val.csv')
df_test = pd.read_csv(cfg.processed_dir / 'tce_test.csv')
print(f'Train: {len(df_train)}, Val: {len(df_val)}, Test: {len(df_test)}')
```

```python
base_pipe = BaselinePipeline()
val_metrics = base_pipe.train(df_train, df_val)
test_metrics = base_pipe.evaluate(df_test)
print('Baseline Test Metrics:', test_metrics['metrics'])
```
